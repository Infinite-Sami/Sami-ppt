Цель работы — построить модель глубокой нейронной сети на базе фреймворка Keras для решения задачи регрессии

  Целевая переменная: `SalePrice` — итоговая стоимость объекта недвижимости в долларах.
  
  Датасет содержит 79 объясняющих признаков (колонок), описывающих различные аспекты жилых домов в Эймсе, штат Айова (площадь, качество материалов, год постройки, количество комнат и т.д.).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')

print("Размер train:", train_df.shape)
print("Размер test:", test_df.shape)
```

### Первичный анализ наборов данных
базовые команды `head`, `tail`, `info`, `describe` и `shape` для тренировочной и тестовой выборок, чтобы оценить структуру данных, типы признаков и наличие пропусков.

```python
print("INFO")
display(train_df.info())
print("\nTRAIN HEAD")
display(train_df.head(3))
print("\nTAIL")
display(train_df.tail(3))
print("\nDESCRIBE")
display(train_df.describe())

print("\nSHAPE", test_df.shape)
```

Удаляем колонку `Id`, так как это порядковый номер, не влияющий на стоимость дома. Объединяем признаки `train` и `test` в общий датафрейм `all_data` и проверяем, сколько строк содержат пропущенные значения (`NaN`).

```python
y_train = train_df['SalePrice']
train_features = train_df.drop(['Id', 'SalePrice'], axis=1)
test_features = test_df.drop(['Id'], axis=1)

ntrain = train_features.shape[0]
all_data = pd.concat([train_features, test_features]).reset_index(drop=True)

rows_with_na = all_data.isna().any(axis=1).sum()
total_na = all_data.isna().sum().sum()

print(f"Всего строк в объединенном датасете: {len(all_data)}")
print(f"Строк, в которых есть хотя бы один пропуск (NaN): {rows_with_na} ({rows_with_na / len(all_data) * 100:.1f}%)")
print(f"Всего пропущенных ячеек в таблице: {total_na}")
print("\nТоп-10 колонок по количеству пропусков:")
display(all_data.isna().sum()[all_data.isna().sum() > 0].sort_values(ascending=False).head(10))
```

Пропуски присутствуют во всех 2919 строках датасета (100% строк имеют хотя бы один `NaN`, чаще всего в признаках бассейна `PoolQC`, аллеи `Alley`, забора `Fence`). Если применить обычное удаление строк с пропусками (`dropna`), мы потеряем все данные.

Поэтому применяем комбинированное заполнение:
1. Там, где `NaN` по документации означает физическое отсутствие постройки (`PoolQC`, `Alley`, `FireplaceQu` и др.), заменяем пропуск на категорию `'None'`.
2. В остальных текстовых признаках заполняем пропуски модой (наиболее частым значением).
3. В числовых признаках заполняем пропуски медианой, так как она устойчива к выбросам.

```python
cat_cols_with_none = ['Alley', 'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1',
                      'BsmtFinType2', 'FireplaceQu', 'GarageType', 'GarageFinish',
                      'GarageQual', 'GarageCond', 'PoolQC', 'Fence', 'MiscFeature']

# Считаем строки и ячейки до каждого этапа заполнения
rows_none = all_data[cat_cols_with_none].isna().any(axis=1).sum()
cells_none = all_data[cat_cols_with_none].isna().sum().sum()
for col in cat_cols_with_none:
    all_data[col] = all_data[col].fillna('None')

obj_cols = all_data.select_dtypes(include=['object']).columns
rows_mode = all_data[obj_cols].isna().any(axis=1).sum()
cells_mode = all_data[obj_cols].isna().sum().sum()
for col in obj_cols:
    all_data[col] = all_data[col].fillna(all_data[col].mode()[0])

num_cols = all_data.select_dtypes(include=['number']).columns
rows_median = all_data[num_cols].isna().any(axis=1).sum()
cells_median = all_data[num_cols].isna().sum().sum()
for col in num_cols:
    all_data[col] = all_data[col].fillna(all_data[col].median())

print(f"1. Заполнено значением 'None': {rows_none} строк ({cells_none} ячеек)")
print(f"2. Заполнено модой (категории): {rows_mode} строк ({cells_mode} ячеек)")
print(f"3. Заполнено медианой (числа): {rows_median} строк ({cells_median} ячеек)")
print(f"Итого осталось пропусков после обработки: {all_data.isna().sum().sum()}")
```

**Промежуточный вывод:**
Все 15 707 пропущенных ячеек во всех 2919 строках успешно заполнены без потери обучающих данных.

Кодирование категорий (One-Hot Encoding)
Переводим категориальные текстовые признаки в числовой бинарный вид с помощью `pd.get_dummies()`, после чего разделяем выборку обратно на `X_train` и `X_test`.

```python
print("Размерность до кодирования:", all_data.shape)
all_data = pd.get_dummies(all_data)
print("Размерность после кодирования:", all_data.shape)

X_train = all_data.iloc[:ntrain, :]
X_test = all_data.iloc[ntrain:, :]
```

**Промежуточный вывод:**
Количество признаков увеличилось с 79 до 301 за счет бинарных колонок для каждой категории.

Масштабирование признаков (StandardScaler)
Применяем `StandardScaler` и сравниваем основные статистические показатели (`mean`, `std`, `min`, `50%`, `max`) на ключевых числовых признаках до и после стандартизации.

```python
from sklearn.preprocessing import StandardScaler

sample_cols = ['LotFrontage', 'LotArea', 'OverallQual', 'YearBuilt', 'GrLivArea']

print("Статистики ДО StandardScaler")
display(X_train[sample_cols].describe().loc[['mean', 'std', 'min', '50%', 'max']].round(2))

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Переводим обратно в DataFrame для наглядного вывода describe()
X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns)

print("\nСтатистики ПОСЛЕ StandardScaler")
display(X_train_scaled_df[sample_cols].describe().loc[['mean', 'std', 'min', '50%', 'max']].round(2))
```

**До масштабирования** признаки имели совершенно разный порядок величин: например, средняя площадь участка `LotArea` составляла **10 516.83** ($\text{std} = 9981.26$), а средняя оценка качества `OverallQual` — всего **6.10** ($\text{std} = 1.38$). Без нормализации признаки с большими абсолютными значениями подавляли бы остальные при градиентном спуске.
**После применения `StandardScaler`** у всех признаков среднее значение (`mean`) стало равно **0.00**, а стандартное отклонение (`std`) приведено к **1.00**. Все признаки теперь находятся в едином масштабе вокруг нуля, что обеспечивает быструю и стабильную сходимость нейронной сети.

### Разделение данных и архитектура нейронной сети
* Делим обучающий набор данных (`X_train_scaled`, `y_train`) в пропорции **80/20** для валидации в процессе обучения.
* Полносвязная нейронная сеть (Sequential) на базе Keras:
  - Входной слой + скрытый слой (128 нейронов, активация `relu`) + Dropout (0.2 для предотвращения переобучения).
  - Скрытый слой (64 нейрона, активация `relu`) + Dropout (0.2).
  - Выходной слой (1 нейрон) для задачи регрессии (предсказание цены).
* Оптимизатор `adam`, функция потерь `mean_squared_error` (MSE), метрика качества `mae`.

```python
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

X_tr, X_val, y_tr, y_val = train_test_split(X_train_scaled, y_train, test_size=0.2, random_state=42)

model = Sequential([
    Dense(128, activation='relu', input_shape=(X_tr.shape[1],)),
    Dropout(0.2),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse', metrics=['mae'])

history = model.fit(
    X_tr, y_tr,
    validation_data=(X_val, y_val),
    epochs=100,
    batch_size=32,
    verbose=1
)
```

### Анализ результатов обучения
Строим график изменения функции потерь на обучающей и валидационной выборках по эпохам

```python
plt.figure(figsize=(10, 5))
plt.plot(history.history['loss'], label='Train Loss (MSE)')
plt.plot(history.history['val_loss'], label='Validation Loss (MSE)')
plt.title('График функции потерь в процессе обучения')
plt.xlabel('Эпоха')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.grid(True)
plt.show()
```

### Оценка качества модели
Считаем метрики MAE и RMSE на валидационном наборе, а также строим график остатков (`Residual Plot`), чтобы проверить наличие систематической ошибки модели.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

val_preds = model.predict(X_val).flatten()

mae = mean_absolute_error(y_val, val_preds)
rmse = np.sqrt(mean_squared_error(y_val, val_preds))

print(f"MAE на валидации: {mae:.2f}")
print(f"RMSE на валидации: {rmse:.2f}")

# Residual Plot (График остатков)
residuals = y_val - val_preds

plt.figure(figsize=(8, 5))
sns.scatterplot(x=val_preds, y=residuals, alpha=0.6)
plt.axhline(y=0, color='r', linestyle='--')
plt.title('График остатков')
plt.xlabel('Предсказанные значения')
plt.ylabel('Остатки (Факт - Прогноз)')
plt.grid(True)
plt.show()
```
